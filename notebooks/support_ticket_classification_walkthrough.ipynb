## Load Run Metadata
Review selected models and test metrics for category and priority tasks.

## Compare Models
Inspect validation performance tables for both outputs.

```python
category_cmp = pd.read_csv(PROCESSED / 'category_model_comparison.csv')
priority_cmp = pd.read_csv(PROCESSED / 'priority_model_comparison.csv')
print('Category models:')
display(category_cmp)
print('Priority models:')
display(priority_cmp)
```

## Error Samples
Review misclassified examples from the test set.

```python
test_preds = pd.read_csv(PROCESSED / 'test_predictions.csv')
cat_errors = test_preds[test_preds['actual_category'] != test_preds['predicted_category']].head(10)
prio_errors = test_preds[test_preds['actual_priority'] != test_preds['predicted_priority']].head(10)
print('Category errors:')
display(cat_errors)
print('Priority errors:')
display(prio_errors)
```
